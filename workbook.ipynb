**Week 4, Day 1 — 📖 Study: What Is a 3D Mesh?**

```python
import numpy as np
import pyvista as pv

# 3 vertices (3D points in space)
vertices= np.array([
    [0.0, 0.0, 0.0],     # vertex 0: origin
    [1.0, 0.0, 0.0],     # vertex 1: 1mm to the right
    [0.5, 1.0, 0.0],     # vertex 2: 0.5mm right, 1mm up
])

# 1 face- connect vertex 0, 1, 2 into a triangle
# The '3' at the start says "Thid face has 3 vertices"
faces= np.array([3, 0, 1, 2])

# Create a PyVista mesh

triangle= pv.PolyData(vertices, faces)  # pv.PolyData(points, faces): A PyVista class object constructor. It takes your raw list of dots (vertices) and your layout map (faces) and bakes them into a valid 3D polygon mesh object.

print("Number of points: ", triangle.n_points)  # triangle.n_points: A built-in PyVista property that automatically counts and returns the number of vertices inside the mesh (will print 3).
print("Number of faces: ", triangle.n_faces)    # triangle.n_faces: A property that counts the number of surface panels inside the mesh (will print 1).
print("Point array: ", triangle.points)    # triangle.points: Returns the internal raw coordinate matrix so you can see the underlying numerical layout.

# Now build something sightly more complex - a cube made of triangles
# This is the kind of structure a bone mesh will have, just wih -9000 faces.
cube= pv.Box()
cube_tri= cube.triangulate()    #.triangulate(): A mesh modification function. It looks at the 6 square faces of the default cube and slices every square diagonally in half to turn them into two triangles. This is critical because medical algorithms like Marching Cubes only speak in triangles.
print(f"\nCube: {cube_tri.n_points} points, {cube_tri.n_faces} faces")


# Visualize both
plotter= pv.Plotter(shape= (1, 2))      # pv.Plotter(...): Creates an interactive window.  shape=(1, 2): A parameter layout instructing the window to split into a grid consisting of 1 row and 2 columns side-by-side.
plotter.subplot(0, 0)
plotter.add_mesh(triangle, show_edges=True, color= "gray")
plotter.add_title("Single triangle")
plotter.subplot(0, 1)
plotter.add_mesh(cube_tri, show_edges= True, color= "gray")
plotter.add_title("Triangulated cube")
plotter.show()
```

**Merging Code for Torso, containing femur head, and Lower Extremity** 

```python
import nibabel as nib
import numpy as np
from scipy.ndimage import zoom

# ── Using the same variables both volumes ──────────────────────────────────────────────
# torso_image torso_data, torso_affine, lower_imgage, lower_data, lower_affine

# ── Extract spacings ───────────────────────────────────────────────
torso_spacing= np.abs([torso_affine[0, 0], torso_affine[1, 1], torso_affine[2, 2]])
lower_spacing= np.abs([lower_affine[0, 0], lower_affine[1, 1], lower_affine[2, 2]])

print(f"Torso spacing= {torso_spacing}")
print(f"Lower extremity spacing= {lower_spacing}")

# ── Resample lower extremity to match torso's X/Y spacing ─────────
# We keep torso as the reference because it has the femoral head
# which is the most critical landmark (C_hip)
# zoom_factors: how much to scale each axis
# For Z: keep as-is (both are 3mm). For X/Y: ratio of spacings

zoom_xy= lower_spacing[0] / torso_spacing[0]    # e.g. 0.76/0.83 ≈ 0.916
zoomr_factors= (zoom_xy, zoom_xy, 1.0)          # only resampling x, and y because z is same in both the scans.

print(f"\nResampling lower extremity X/Y by factor: {zoom_xy:.4f}")
print("This may take 30-60 seconds")

lower_resampled= zoom(lower_data, zoomr_factors, order=1)

# order= 1 bilinear interpolation - good balance and quality
# order= 0 would be nearest neighbour (faster but blocky)
# order= 3 would be cubic (slow, unnecessary for bone CT)

print(f"Lower resampled shape: {lower_resampled.shape}")
print(f"Torso shape: {torso_data.shape}")
# X and Y dimensions should match torso's 512x512

# ── Handle different X/Y sizes after resampling ───────────────────
# After zoom, X/Y may not be exactly 512, Pad or crop to match.
target_xy= torso_data.shape[0]
current_xy= lower_resampled.shape[0]

if current_xy != target_xy:
    diff= target_xy - current_xy
    if diff > 0:
        # Pad with background HU value (-1024 = air)
        pad= diff // 2
        pad_extra= diff - pad * 2
        lower_resampled= np.pad(
            lower_resampled,
            ((pad, pad + pad_extra), (pad, pad + pad_extra), (0, 0)),
            mode= "constant", constant_values= -1024
        )
        
    else:
        # Crop from center
        start= (-diff) // 2
        lower_resampled= lower_resampled[:start+target_xy, start:start+target_xy, :]
        print(f"Lower after size match: {lower_resampled.shape}")

# ── Find the overlap and split point ──────────────────────────────
# Torso: Z origin = -719mm, Z end = +13 mm
#Lower: Z origin = -1597.5mm, Z end= -682.5mm
# We concatenate: all of ower extremity + the torso slices that are BELOW the lower extremity's top (i.e. below -682.5mm)

torso_z_origin= torso_affine[2, 3] # -719.0 mm
lower_z_end= lower_affine[2, 3] + lower_data.shape[2] * lower_affine[2, 2]
z_step= torso_affine[2, 2]

print(f"\nTorso Z origin:    {torso_z_origin:.1f} mm")
print(f"Lower Z end:       {lower_z_end:.1f} mm")
print(f"Z step:            {z_step:.1f} mm")            # z_step: Confirms the slice thickness step resolution (3.0 mm).


# how many torso slices sit below the lower extremity's top?
# These slices have Z positions from torso_z_origin up to lower_z_end
# We want torso slices starting from the point where lower extremity ends
overlap_slices= int(round((lower_z_end - torso_z_origin) / z_step))
print(f"Torso Slices to skip (in overlap zone): {overlap_slices}")

# Take only the torso slices that are above the lower extremity
#i.e., slices from index "overlap_slices" onward (towar the head)
torso_above_knee= torso_data[:, :, overlap_slices:]
print(f"Torso portion above lower extremity: {torso_above_knee.shape}")

# ── Concatenate: lower (feet→knee) + torso_above (knee→head) ──────
# Stack along Z axis (axis=2)
merged_data= np.concatenate([lower_resampled, torso_above_knee], axis=2)
print(f"\nMerged volume shape: {merged_data.shape}")

coverage_cm= merged_data.shape[2] * abs(z_step) / 10
print(f"TOtal Z coverage: {coverage_cm:.1f} cm")

# ── Build the affine for the merged volume ─────────────────────────
# The merged volume starts where the lower extremity starts (the feet)
# and uses the torso's X/Y spacing as the reference

merged_aff= torso_affine.copy()
merged_aff[0, 0]= torso_spacing[0]      # X spacing
merged_aff[1, 1]= torso_spacing[1]      # Y spacing
merged_aff[2, 2]= z_step                # Z spacing (3mm)
merged_aff[2, 3]= lower_affine[2, 3]    # Z origin= wehre lower extremity starts (feet)

print(f"\nMerged affine Z origin: {merged_aff[2, 3]:.1f} mm (should be ≈ -1597.5)")

# ── Save the merged volume ─────────────────────────────────────────
merged_img= nib.Nifti1Image(merged_data.astype(np.int16), merged_aff)
nib.save(merged_img, "./data/raw/patient_01_merged.nii.gz")
print("\nMreged Volume saved to: ./data/raw/patient_01_merged.nii.gz")

# ── Final verification ─────────────────────────────────────────────
verify = nib.load("./data/raw/patient_01_merged.nii.gz")
v_data= verify.get_fdata()
v_spacing= np.abs([verify.affine[0, 0], verify.affine[1, 1], verify.affine[2, 2]])
print("Verification: ")
print(f"Shape: {v_data.shape}")
print(f"Spacing: {v_spacing}")
print(f" HU Range: {v_data.min()} to {v_data.max()}")
print(f" Coverage: {v_data.shape[2] * v_spacing[2] / 10:.1f} cm")
```

**Week 4, Day 2 — 📖 Study: The Marching Cubes Algorithm**

```python
import numpy as np
import pyvista as pv
from skimage.measure import marching_cubes

# Recreate the synthetic bone mask from Week 2 Day 4

volume= np.zeros((100, 100, 100), dtype= np.uint8)
xx, yy, zz= np.meshgrid(range(100), range(100), range(100), indexing="ij")
dist= np.sqrt((xx-50)**2 + (yy-50)**2 + (zz-50)**2)
volume[dist < 20] = 1  # sphere of radius 20 voxels

print("Bone mask shape= ", volume.shape)
print("Bone voxels= ", volume.sum())

# Run MArching Cubes
# spacing= physical size of each voxel in mm
# For this synthetic data, pretend voxel is 1x1x1 mm

voxel_spacing= (1.0, 1.0, 1.0)

vertices, faces, normals, values = marching_cubes(
    volume,
    level = 0.5,
    spacing= voxel_spacing
) 

print(f"\nMarching Cube output: ")
print(f"Vertices: {vertices.shape}")        #(N, 3) - N 3D points
print(f"Faces: {faces.shape}")              #(M, 3) - M triangles, each referencing 3 vertices

# Convert to PyVista format
# PyVista needs faces as [3, v0, v1, v2, v3, v0, v1, v2, ...]
faces_pv = np.hstack([
    np.full((len(faces), 1), 3, dtype=int),
    faces
])
mesh= pv.PolyData(vertices, faces_pv)

print(f"\nPyVista mesh: ")
print(f" Points: {mesh.n_points}")
print(f" faces:  {mesh.n_faces}")

#Visualize- should look like a sphere with staircase artifacts
plotter= pv.Plotter()
plotter.add_mesh(mesh, color="gray", show_edges=True)
plotter.add_title("Synthetic bone - raw marching Cubes")
plotter.show()
```

```python
import nibabel as nib
import numpy as np
import pyvista as pv
from skimage.measure import marching_cubes
from pathlib import Path

# ── Load the total task segmentation ──────────────────────────────
seg_img= nib.load("./data/segmentations/patient_01_total.nii.gz")
seg_data= seg_img.get_fdata().astype(int)
affine= seg_img.affine

print("Segmentation shape= ", seg_data.shape)
print("All labels present= ", np.unique(seg_data))

# ── Extract femur_left binary mask ─────────────────────────────────
# REPLACE the number below with your actual femur_left label ID
FEMUR_LABEL = 76

binary_mask= (seg_data == FEMUR_LABEL).astype(np.uint8)         # seg_data==FEMUR_LABEL outputs only the voxles that are of femur, .astype(int)= converts to integer
voxel_count= binary_mask.sum()                                  # binary_mask.sum() adds all the voxels data
print(f"\nFemur Label {FEMUR_LABEL}: {voxel_count} voxels")

if voxel_count == 0:
    print("ERROR: No voxel found. Please check your label ID.")
else:
    # ── Extract voxel spacing from affine ──────────────────────────
    voxel_spacing= tuple(np.abs([affine[0, 0], affine[1, 1], affine[2, 2]]))
    print(f"Voxel spacing= {voxel_spacing} mm")
    
    # ── Run Marching Cubes ─────────────────────────────────────────
    print("\nRunning Marching Cubes...")
    vertices, faces, normals, values = marching_cubes(
        binary_mask,
        level= 0.5,
        spacing= voxel_spacing
    )

    print(f"Vertices: {len(vertices)}")
    print(f"Faces: {len(faces)}")

    # ── Convert to PyVista ─────────────────────────────────────────
    faces_pv= np.hstack([
        np.full((len(faces), 1), 3, dtype= int), faces
    ])
    
    # len(faces)= counts the triangles in the 3D mesh.  np.full((..., 1), 3, dtype=int)= creates a single, vertical column filled with entirely the number 3 absed on the number of rowns in faces. If you ahve 1000 faces, it amkes a column of 1000 threes.

    #np.hstack([])= Horizontal stack works like digital glue. It places the new column of 3 side by side with the original faces data.

    raw_mesh= pv.PolyData(vertices, faces_pv)

    # ── Save ───────────────────────────────────────────────────────
    Path("./meshes/").mkdir(exist_ok=True)
    raw_mesh.save("./meshes/femur_raw.stl")
    print(f"\nRaw mesh saved. Faces {raw_mesh.n_faces}")

    # ── Visualize ──────────────────────────────────────────────────
    plotter = pv.Plotter()
    plotter.add_mesh(raw_mesh, color='#E8C9A0', opacity=0.9)
    plotter.add_title(f"Real Human Femur — Raw ({raw_mesh.n_faces} faces)")
    plotter.show()    
```

**Marching cubes on appendicualar**

```python
import nibabel as nib
import numpy as np
from skimage.measure import marching_cubes
import pyvista as pv
from pathlib import Path

app_img = nib.load("./data/segmentations/patient_01_appendicular.nii.gz")
app_data = app_img.get_fdata().astype(int)
app_aff= app_img.affine

# Replace with the actual tibia label ID
TIBIA_LABEL= 2 

tibia_mask= (app_data == TIBIA_LABEL).astype(np.uint8)
print(f"Tibia label {TIBIA_LABEL}: {tibia_mask.sum()} voxels")      #np.sum(tibia_mask) will aslo give the same output as tibia_mask.sum()

app_spacing= np.abs([app_aff[0, 0], app_aff[1, 1], app_aff[2, 2]])

# Extract surface via marching cubes
verts_t, faces_t, _, _ = marching_cubes(
    tibia_mask,
    level= 0.5,
    spacing= app_spacing,
)

faces_t_pv= np.hstack([
    np.full((len(faces_t), 1), 3, dtype=int), faces_t
])
tibia_raw= pv.PolyData(verts_t, faces_t_pv)
tibia_raw.save("./meshes/tibia_raw.stl")
print(f"Tibia raw mesh saved, Faces= {tibia_raw.n_faces} & No. of points= {tibia_raw.n_points}, points= {tibia_raw.points}")

# ── Visualize both bones together ─────────────────────────────────
plotter = pv.Plotter()
plotter.add_mesh(raw_mesh,   color='#E8C9A0', opacity=0.9, label='Femur')
plotter.add_mesh(tibia_raw, color='#C8D8E8', opacity=0.9, label='Tibia')
plotter.add_legend()
plotter.add_title("Femur + Tibia — Raw Meshes")
plotter.show()
```

**Code to separate the tibias into separate left and right tibia**

```python
import pyvista as pv

# Split the double-tibia mesh into two separate objects
tibia_bodies= tibia_raw.split_bodies()      #.split_bodies()= looks for floating, disconnected pieces in a mesh and breaks them into separate PloyData objects.

print(f"Number of tibia meshes found: {len(tibia_bodies)}")

# Get teh X-center coordinates of the femur mesh
femur_x_center= raw_mesh.center[0]          # .center[]= calculates the bounding bo center (x, y, z) and the index[0] extracts x-coordinate (left/right axis)
print(f"Femur X-center: {femur_x_center:.2f}")

# Check each tibia's X-center to see which one aligns with the femur
for i, t_mesh in enumerate(tibia_bodies):
    tibia_x_center = t_mesh.center[0]
    print(f"Tibia body {i} X-center: {tibia_x_center:.2f}")

# Programmatically pick the matching Tibia (closet X coordinate to the femur)
matching_tibia= min(tibia_bodies, key=lambda m: abs(m.center[0] - femur_x_center))
other_tibia= max(tibia_bodies, key=lambda m: abs(m.center[0] - femur_x_center))

# Convert the UNSTRUCTURED GRID to PolyData object before saving as .stl
matching_tibia_poly= matching_tibia.extract_surface()
other_tibia_poly= other_tibia.extract_surface()

# Save them cleanly n the meshes folder
matching_tibia_poly.save("./meshes/tibia_left_raw.stl")
other_tibia_poly.save("./meshes/tibia_right_raw.stl")

print("Saved matching tibia as 'tibia_left-raw.stl'!")
```

**Day 5 — 💻 Code: Smooth and Decimate Both Bones**

```python
import pyvista as pv
import nibabel as nib

# ── Load raw meshes saved on Day 3 ────────────────────────────────
femur_raw = pv.read("./meshes/femur_left_raw.stl")
tibia_raw  = pv.read("./meshes/tibia_left_raw.stl")

# Print the faces in raw bones
print(f"Femur raw: {femur_raw.n_faces} faces")
print(f"Tibia raw: {tibia_raw.n_faces} faces")

# ── Process femur ──────────────────────────────────────────────────
femur_smooth= femur_raw.smooth(n_iter= 30, relaxation_factor= 0.1)
femur_decimate= femur_smooth.decimate(target_reduction= 0.8)
femur_final= femur_decimate.compute_normals(auto_orient_normals= True)

# ── Process tibia ──────────────────────────────────────────────────
tibia_smooth= tibia_raw.smooth(n_iter= 30, relaxation_factor= 0.1)     #.smooth() runs Laplacian smoothing 
tibia_decimate= tibia_smooth.decimate(target_reduction= 0.8)       #.decimate= reduces the faces in the 3D model.
tibia_final= tibia_decimate.compute_normals(auto_orient_normals= True)


# ── Print reduction summary ────────────────────────────────────────
print(f"\nFemur: {femur_raw.n_faces} → {femur_final.n_faces} faces", f"{(femur_raw.n_faces - femur_final.n_faces) / femur_raw.n_faces *100}% reduction")
print(f"Tibia: {tibia_raw.n_faces} → {tibia_final.n_faces}", f"{(tibia_raw.n_faces - tibia_final.n_faces) / tibia_raw.n_faces * 100}% reduction")

# ── Save processed meshes ──────────────────────────────────────────
femur_final.save("./meshes/femur_left_processed.stl")
tibia_final.save("./meshes/tibia_left_processed.stl")

# ── Visualize 3 stages for femur & tibia────────────────────────────
plotter= pv.Plotter(shape=(2, 3))

plotter.subplot(0, 0)
plotter.add_mesh(femur_raw, color="#E8C9A0", show_edges= True)
plotter.add_title(f"Raw\n{femur_raw.n_faces} faces")

plotter.subplot(0, 1)
plotter.add_mesh(femur_smooth, color= "#E8C9A0", show_edges= True)
plotter.add_title(f"Smoothed\n{femur_smooth.n_faces} faces")

plotter.subplot(0, 2)
plotter.add_mesh(femur_final, color="#E8C9A0", show_edges= True)
plotter.add_title(f"Final\n{femur_final.n_faces} faces")

plotter.subplot(1, 0)
plotter.add_mesh(tibia_raw, color="#E8C9A0", show_edges= True)
plotter.add_title(f"Raw\n{tibia_raw.n_faces} faces")

plotter.subplot(1, 1)
plotter.add_mesh(tibia_smooth, color="#E8C9A0", show_edges= True)
plotter.add_title(f"Smoothed\n{tibia_smooth.n_faces} faces")

plotter.subplot(1, 2)
plotter.add_mesh(tibia_final, color="#E8C9A0", show_edges= True)
plotter.add_title(f"Final\n{tibia_final.n_faces} faces")

plotter.show()
```
